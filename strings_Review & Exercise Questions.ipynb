# Strings and string manipulation

### Creating Strings

 - `""`
 - `''`
 - `\`
 - `"""`
 - `'''`
 - `\'`
 - `\"`


```python
# Single-line strings
single_quotes = 'This is a single-quoted string.'
double_quotes = "This is a double-quoted string."
backslash = 'This is a string with a backslash: \\'

# Multi-line strings using triple quotes
multi_line_single_quotes = '''This is a multi-line
single-quoted string.'''
multi_line_double_quotes = """This is a multi-line
double-quoted string."""

# Quotes inside strings
single_quotes_in_string = "I'm using single quotes inside a double-quoted string."
single_quotes_in_string_alt = 'I\'m using single quotes inside a single-quoted string.'
double_quotes_in_string = 'He said, "Hello!"'
double_quotes_in_string_alt = "He said, \"Hello!\""

print(single_quotes)
print(double_quotes)
print(backslash)
print(multi_line_single_quotes)
print(multi_line_double_quotes)
print(single_quotes_in_string)
print(single_quotes_in_string_alt)
print(double_quotes_in_string)
print(double_quotes_in_string_alt)

```

### Operators

- `+` = concatenation
- `*` = concatenates multiples copies of a string
- `in`/`not in` = membership

```python
s = "This  : a ,, string ! with some ; punctuation."

# Concatenation
s1 = s + " Concatenated."
print(s1)

# Repetition
s2 = s[0:4] * 3
print(s2)

# Membership
print("string" in s)
print("string" not in s)
print("desolate" not in s)
print("desolate" in s)
```

### Indexing and Slicing

 - single character
 - range of characters
 - reverse a string
 - change step size

```python
s = "Hello, World!"

# Select a single character
print(s[0]) 

# Select multiple characters
print(s[0:5])  

# Reverse a string
print(s[::-1])  

# Slice with a step size of 2
print(s[::2])  

# Slice with a step size of 3
print(s[::3])  


```

### Modifying strings

- immutable
- slicing and concatenation
- `.replace()`

```python
s = "Hello, World!"

# Trying to modify the string directly will raise an error
# s[0] = 'h'  # Uncomment this line to see the error

# Creating a new string using slicing
new_slicing = s[:6] + 'Python!'
print(new_slicing)  

# Creating a new string using concatenation
new_concatenation = s + ' I love programming.'
print(new_concatenation)  

# Creating a new string using replace()
new_replace = s.replace('World', 'string methods')
print(new_replace) 
```
