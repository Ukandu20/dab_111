### Docstrings 

Review the documentation at [PEP257](https://peps.python.org/pep-0257/) for full details.

To create a docstring: 
 - always use `"""triple double quotes"""` 
     - use `r"""raw triple double quotes"""` if you use any backslashes

There are two types:
 - one-line docstrings
 - multi-line docstrings

#### One-line docstrings

- triple quotes are used even though the string fits on one line. This makes it easy to later expand it.
    - one line is a maximum of 72 characters
- closing quotes are on the same line as the opening quotes. This looks better for one-liners.
- no blank line either before or after the docstring.
- the docstring is a phrase ending in a period. It prescribes the function or method’s effect as a command (“Do this”, “Return that”), not as a description
- the nature of the return value cannot be determined by introspection, so it should be mentioned. The preferred form for such a docstring would be something like:

```python
def function(a, b):
    """Do X and return a list."""
    <statement(s)>
```

#### Multi-line docstrings 

 - consist of a summary line just like a one-line docstring, followed by a blank line, followed by a more elaborate description. 
     - important that it fits on one line and is separated from the rest of the docstring by a blank line 
     - the summary line may be on the same line as the opening quotes or on the next line
 - summarize function behavior
 - document:  
     - arguments
     - return value(s)
     - side effects 
     - exceptions raised 
     - restrictions on when it can be called (all if applicable)
     - optional arguments should be indicated 
     - whether keyword arguments are part of the interface

Create a function that will take in the following arguments:
- name of a text file
- whether punctuation should be removed or not
    - make sure this has a default value to not remove puntuation
- whether or not to count words
    - make sure this has a default value to not count words

Your function should:
 - assert that the filename is a string
 - assert that the extension of the filename is ".txt"
 - use `try`/`except` to catch a FileNotFoundError when trying to read in the file
     - if no error, read the text in as a string
 - lower case the text
 - check punctuation argument
     - if True, remove punctuation from provided string
 - check word count argument
     - if True, calculate total word count and unique word count
 - return text as a list split on white space and word counts if necessary
 

````python
import string

def remove_punct(text):
    """
    Remove all punctuation characters from the input text.

    Parameters:
    text (str): The input string from which punctuation will be removed.

    Returns:
    str: A new string with all punctuation characters removed.
    """
    text_no_punct = ""
    for ch in text:
        if ch not in string.punctuation:
            text_no_punct += ch
            
    return text_no_punct
````

````python
def word_counts(lst):
    """
    Count the total number of words and the number of unique words in a list.

    Parameters:
    lst (list of str): A list of words (strings).

    Returns:
    tuple: A tuple containing:
        - total (int): Total number of words in the list.
        - unique (int): Number of unique words in the list.
    """
    total = len(lst)
    unique = len(set(lst))
    return total, unique
````

````python
def get_text_data(fname, punct=False, count_words=False):
    """
    Read text data from a file, optionally remove punctuation, and optionally count words.

    Parameters:
    fname (str): The filename of the .txt file to read.
    punct (bool): If True, removes punctuation from the text. Default is False.
    count_words (bool): If True, returns total and unique word counts. Default is False.

    Returns:
    list: A list of words in the file (processed based on flags).
    tuple: If count_words is True, returns (total_word_count, unique_word_count, word_list).

    Raises:
    AssertionError: If fname is not a string or does not end with '.txt'.
    Prints an error message if the file is not found.
    """
    assert isinstance(fname, str), "filename needs to be a string"
    assert fname[-4:]==".txt", "file needs to be a text file"
    try:
        with open(fname) as f:
            data = f.read()
    except FileNotFoundError:
        print("no such file")
        
    data = data.lower()
    
    if punct:
        data = remove_punct(data)
        #return data

    data_list = data.split()
        
    if count_words:
        total, unique = word_counts(data_list)
        #return total, unique
    #else:
        #return data

    return data, total, unique

    
````
