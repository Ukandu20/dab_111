>Write a program n that takes a list as an user input and extracts a new list containing only the unique even numbers in the input list, sorted in ascending order. The `sorted()` function may be helpful. 

```python
2121,123,1,32,245,12123221,13222,4534675665,34345678,23312,22,44
```

```python
userList = (input("Enter a list of numbers")).split(',')
evenList = []

for i in userList:
    number = int(i.strip())
    if number %2==0 and number not in evenList:
            evenList.append(number)
            evenList = sorted(evenList)

evenList
```
