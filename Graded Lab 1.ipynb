## Question 1 (5 Marks)
**Write a Python program that:**

- Creates two lists, each containing 15 random integers.
    - The first list should have values between **100 and 300**.
    - The second list should have values between **200 and 500**.
- Use random.seed(<<Group#>>) before generating list to ensure reproducibility.
- Add both lists element-wise to form a third list (sum list).
- Your program should print the following as output.

    - First list
    - Second list
    - Sum list
    - Maximum number in the sum list

**Sample code to generate list**

````
random.seed(10) # If your group number is 10
list1 = [random.randint(100, 300) for _ in range(15)]
````

`````python
# Question 1 Code must be added here
# Import the module used to generate random integers.
import random

# Function to generate a random list.
def generate_rnd_list(a,b,length):
    # Initialize an empty list.
    list = []

    # Reset the seed so calls with the same arguments reproduce the same list.
    random.seed(42)
    for i in range(length):
        
        # Choose a random integer and add it to the result list.
        rnd_num = random.randint(a,b)
        list.append(rnd_num)
    # Return the completed list.
    return list


# Generate two lists of 15 integers using the specified ranges.
list1 = generate_rnd_list(100, 300, 15)
list2 = generate_rnd_list(200, 500, 15)

# Store the sums of matching positions in a third list.
list3 = []

# Find the lengths of both lists.
x= len(list1)
y= len(list2)

# Only combine the lists when they contain the same number of elements.
if x == y:
    for i in range(x):
        # Add the values at the current index and save the sum.
        sum = list1[i] + list2[i]
        list3.append(sum)

# Find the largest value in the new sum list.
max_list3 = max(list3)


# Display both original lists, their sums, and the largest sum.
print(f"my generated list1 is: {list1}")
print(f"my generated list2 is: {list2}")
print(f"my generated list3 is: {list3}")
print(f"The largest number in my summed list is: {max_list3}")
`````

## Question 2 (5 Marks)

**Write a Python program that:**

- Prompts the user to enter how many people records they want to input.
- For each person, ask the user to provide:
    - name (string)
    - age (integer)
    - city (string)
- Store each person's information in a dictionary with the keys: "name", "age", and "city".
- Add each dictionary to a list.
- After all entries are collected, print the final list.

`````python
# Question 2 Code must be added here


# Determine how many person records to collect and convert the count to an integer.
userCount = int(input("Enter the total number of records allowed"))

# Initialize the list that will hold all person records.
list1 = []
# Collect one record per iteration.
for i in range(userCount):

    # Create a new dictionary for this person.
    users = {}
    # Collect each field, converting the age to an integer.
    users['name'] = str(input("Enter your name: "))
    users['age']  = int(input("Enter your age: "))
    users['city'] = str(input("What city are you from?: "))

    # Add the completed person record to the list.
    list1.append(users)


# Display all collected records.
print(list1)
`````

## Question 3 (5 Marks)

**Write a Python program that:**

- Prompts the user to enter a list of comma-separated values (e.g., 45, apple, 12, 67, 45, banana, 100).
- Converts the input into a list.
- From this list:
    - **Remove all duplicate** values.
    - **Eliminate all non-numerical entries** (anything that is not a number).
    - Convert **valid numerical values** into **integers**.
- Print the cleaned list in ascending order.

`````python
# Question 3 Code must be added here

# Split the input at commas to create a list of strings.
userList = input("Enter the comma-separated list").split(",")
# Initialize a placeholder for the cleaned values.
cleanedList = []
# Display the original values and the type of the container.
print(f"This is the original user entered list: ", userList)
print(type(userList))


# Strip surrounding spaces and remove duplicate strings using a set.
cleanedList = list(set(element.strip() for element in userList))
print(f"This is the cleaned list without duplicates: ",cleanedList)

# Collect integer values from the cleaned strings.
numericList = []
for i in cleanedList:
    # Keep digit-only strings.
    if i.isdigit():
        # Convert the accepted string to an integer and store it in numericList.
        numericList.append(int(i))

# Display the sorted numeric values.
print(f"This is the numeric list: ",sorted(numericList))   

# Confirm the type of each converted value.
for element in numericList:
    print(type(element)) 
`````

# Question 4 (5 Marks)

You are provided with a CSV file named *question4.csv* that contains a header row followed by 3 rows of data. 
    
**Write a Python program to:**

- Open and read the data.csv file using the csv module.
- Use the first row as the keys of the dictionary (i.e., column headers).
- Store each of the 3 records (rows) as a dictionary using those keys.
- Add each dictionary to a list.
- Print the final list of dictionaries.

`````python
import csv

# Open the CSV file for reading.
with open('question4.csv') as f:

    # Use the header row as dictionary keys for each subsequent data row.
    q4_csv = csv.DictReader(f)
    # Collect the row dictionaries in a list.
    q4_list = []
    # Read each data row and append its dictionary to the list.
    for row in q4_csv:
        q4_list.append(row)
        
# Display the complete list of records.
print(q4_list)
`````

## Question 5 (5 Marks)

**Write a Python program that asks the user to enter their email address and password, and checks the validity of both based on the following criteria:**

- *Email Validation Rules*
    - The email must contain an @ symbol.
    - The email must contain a .com or .edu or .org domain.
    - The email must be at least 6 characters long.
    
      
- *Password Validation Rules*
    - The password must be at least 8 characters long.
    - It must contain at least one letter and at least one digit.
    - It should not contain spaces.
  

- *Output Requirements*
    - If both email and password are valid → Print "Login credentials accepted."
    - If either email or password is invalid → Print appropriate error messages (e.g., "Invalid email format.", "Password too weak.")

`````python
# Read the email address and password as strings.
userEmail = input("Enter a valid email:")
userPassword = input("Enter a valid password: ")


# Define the allowed email endings without their leading dots.
validDomains = ['edu', 'com', 'org']

#  Email validation rules to check email length, the presence of @, and the domain name after the last dot using reverse partitioning. 
emailLength, hasAtSymbol, domainName = len(userEmail) >= 6, "@" in userEmail, any(userEmail.rpartition('.')[-1] in domain for domain in validDomains)

# Password validation rules to check for at least eight characters, no literal spaces, a digit, and a letter.
passwordLength, noSpace, hasDigit, hasAlpha = len(userPassword) >= 8, " " not in userPassword, any(char.isdigit() for char in userPassword), any(char.isalpha() for char in userPassword)


# A valid email requires all three email checks to pass.
validEmail = (
    emailLength
    and hasAtSymbol
    and domainName
)

# A valid password requires all four password checks to pass.
validPassword = (
    passwordLength 
    and noSpace 
    and hasDigit
    and hasAlpha
)

# Accept the credentials only when both sets of checks pass.
if validEmail and validPassword:
    print("Login credentials accepted.")

# Report an email error first if the email check fails.
elif not validEmail:
    print(userEmail)
    print("Error, Invalid email format.")
    
# If the email passes but the password fails, report a weak password.
else:
    print("Password too weak")
`````
