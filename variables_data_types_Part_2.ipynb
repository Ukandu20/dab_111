### Mapping types: `dict`

#### Dictionary 

Lists may be constructed in several ways:

 - using `key:value` pairs with braces:
    - empty list: `{}`
    - separating items with commas: `{'a':1, 'b':2, 'c':3}`
 - dictionary comprehension: 
    - `{x:x**2 for x in range(10)}`
 - type constructor: 
    - empty list: `dict()` 
    - from iterable: `dict([('a', 1), ('b', 2), ('c', 3)])`, `dict(a=1, b=2, c=3)`

```python
# create some dictionaries
dict1 = {'a': 1, 'b': 12, 'c': 35}
dict2 = dict([('a', 12), ('b', 98), ('c', 9877)])
```

#### Dictionary operations

Dictionaries various [methods/operations](https://docs.python.org/3/library/stdtypes.html#dict) as noted in the docs.

##### Dictionary  properties

The important properties of dictionaries are as follows:
 - access values by keys
 - iterable
 - mutable
 - unordered
 - keys must be hashable
 - can be nested
 - dynamic


```python
# access via key
dict1['a']
```

```python
# mutable: d[k]=v, del, update'
dict1['b'] = 311
dict1
```

```python
# nesting
dict1.values()
```

```python
# iterable
for key, value in dict1.items(): #returns keys and values
    print(f"This is a key, value pair:  key: {key}, Value: {value}")
for key in dict1.keys(): #returns keys
    print(f"This is one of the keys: ",key)

for value in dict1.values(): #returns values
    print(f"This is one of the values: {value}")

```

### Six Additional methods for Dicts

```python
drivers = {'user1': 'James', 'user2': 'Earl', 'user3': 'John', 'user4': 'Frank'}
carBrands = {'1001A': 'Tesla', '1001B': 'Buick', '1001C': 'Ram', '1001D': 'Ford'}
```

```python
#Method 1: copy() Returns a copy of the dictionary. It also preserves the original dictionary from being accidentally altered
owners = drivers.copy()
print(f"This is a copy of the original `drivers` dictionary, now known as `owners`: ", owners)

#Method 2: clear() clears all elements from a dictionary and returns it as empty

resetDrivers = drivers.clear()
print(f"This is the empty `drivers` dictionary: ", resetDrivers)

#Method 3: list() returns a list containing the keys that are in the dictionary
carIds = list(carBrands)
print(f"These are the car Ids: ", carIds)

#Method 4: reversed() This reverses the order of which the items in the dictionary are presented
driversReversed = dict(reversed(owners.items()))
print(f"This is the dictionary of drivers in reversed order:", driversReversed)

#Method 5: len() returns the length of the dictionary items
carsLength = len(carBrands)
print(f"There are {carsLength} cars in our portfolio")

#Method 6: update() updates the value of a specific key in the dictionary
owners.update({'user3': 'Michael'})
print(f"user3's value has been updated in the dictionary: {owners}")
```
