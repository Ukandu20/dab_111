Write a class called 'Inventory' with attributes 'item_id', 'item_name', and 'count'. It should have two methods: 'remove' and 'receive', both of which correctly update the 'count' attribute. In addition, 'remove' should print "Not enough items availabe." if the number requested exceeds the available count. Submit only the code that defines the class.

```python
class Inventory:
    def __init__(self, item_id, item_name, count):
        self.item_id = item_id
        self.item_name = item_name
        self.count = count

    def remove(self, amount):
        if amount> self.count:
            print("Not enough items available")
        else:
            self.count = self.count - amount
            return self.count

    def receive(self, amount):
        self.count = self.count + amount
        return self.count

```

```python
item_1 = Inventory(1,'Fish', 10)

item_1.remove(1)
item_1.receive(12)
```

```python
from abc import ABC, abstractmethod

class Students(ABC):
    def __init__(self, student_id, fName, lName):
        self.student_id = student_id
        self.student_fname = fName        
        self.student_lname = lName

    @abstractmethod
    def student_info(self):
        print(f"Student Id: {self.student_id} \n Student Name: {self.student_fname} {self.student_lname} \n")


class Grades(Students):
    def __init__(self, student_id, fName, lName, assignment, midterm, final):
        super().__init__(student_id, fName, lName)
        self.assignment_score = assignment
        self.midterm_score = midterm
        self.final_score = final

    def student_info(self):
        self.final_grade = (((self.assignment_score / 100) * 0.4) + ((self.midterm_score / 100) * 0.3) + ((self.final_score / 100) * 0.3)) * 100
        super().student_info()
        print(f"{self.final_grade}")
```

```python
students = []
student_count = int(input("How many students? "))

for i in range(student_count):
    print(f"\nStudent {i + 1}")

    student_id = int(input("Student ID: "))
    first_name = input("First name: ")
    last_name = input("Last name: ")
    assignment = float(input("Assignment score: "))
    midterm = float(input("Midterm score: "))
    final = float(input("Final exam score: "))

    student = Grades(
        student_id, first_name, last_name,
        assignment, midterm, final
    )
    students.append(student)

for student in students:
    student.student_info()
```

Create a function called is_palindrome that takes a string as an argument and returns True if the string reads the same forward and backward, and False otherwise. Ignore spaces and capitalization.

Example output on implementing function:

- print(is_palindrome("racecar"))        # True
- print(is_palindrome("hello"))          # False
- print(is_palindrome("Never odd or even"))  # True
- print(is_palindrome("Python"))         # False

```python
def is_palindrome(x):
    assert type(x) is str, "x has to be a string"
    x_reverse = ''
    i = 0
    x = x.lower().replace(' ', '')
    j = len(x)
    for i in range(j):
        x_reverse += x[j - 1]
        j = j - 1
    if x_reverse == x:
        return True
    else:
        return False

is_palindrome('Never odd or even')
```
