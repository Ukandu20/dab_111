```python
common = []
count = 0
def common_digits(x,y):
    """ 
            Takes two integers and returns the number of common digits between them.
        
            Parameters: 
                x(int): Integer 1
                y(int): Integer 2
                
            Returns:
                int: Returns the value of the total number of common digits
                
    """
    i = 0
    min_value = min(x,y)
    for i in range(min_value + 1):
        j = i + 1
        if x % j == 0 and y % j == 0:
            common.append(j)
    count = len(common)
    return common, count
    

common_factors, common_factors_count = common_digits(456, 564)
print(common_factors)
print(common_factors_count)
```

```python
list2 = ['agddg', 'ae', 'wqwqwefwefdhf', 23, 232324.53]

def conv_to_str(x=[]):
    """ 
            Purpose: 
                converts every element of a list into a string
    
            Parameters/Arguments: 
                x(list): which is the list of integers that will be checked. It defaults to an empty list
    
            Return: 
                int: The largest consecutive sum, or 0 if no pair exists. 
    """
    list_str = []
    for element in x:
        element = str(element)
        list_str.append(element)
    return list_str
        
def len_of_str(x=[]):
    """ 
        Converts all items in a list to strings and returns the length of the strings.
    
        Parameters: 
            x(string-list): which is the list of items of varying datatypes. It defaults to an empty list
            
        Returns:
            list: Returns a list containing the length of every element from the orginal list
            
    """
    str_len = []
    for element in conv_to_str(x):
        str_len.append(len(element))

    return str_len

def average_str_length(x=[]):
    assert isinstance(x, list), "x must be a list."
    count = len(x)
    sum = 0

    if count > 0:
        for i in len_of_str(x):
            sum = sum + i
        average = round(sum / count)
    else:
        print("List is empty")

    return average

average = average_str_length(list2)
print(f"The average string length is approximately {average}")
    
```
