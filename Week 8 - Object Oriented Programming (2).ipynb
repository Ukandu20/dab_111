#Object Oriented Programming


Object Oriented Programming is not	a language but a **programming paradigm**

Object-Oriented Programming (OOP) encapsulates attributes (data) and behaviors (methods) within a single entity, ensuring better organization and security.

**Abstraction** is a process where you only show the data that is relevant to the user
Any other information is hidden away

**Encapsulation** enhances data integrity by restricting direct access and enforcing controlled interactions.

**Inheritance** allows an object to inherit (share) attributes and methods from another object. Inheritance is a concept that allows for the code reuse and better overall design

**Polymorphism** allows different classes to use the same method name but with different implementations, enabling flexibility.

     In python dynamic binding fetaure determines method calls at runtime, allowing for flexibility in method execution.

# Inheritence

**Vehicle class**
- Create a parent class Vehicle that includes common attributes (brand, model, year).
- Implement a method show_details() to display vehicle information.

**Car Class**
- Inherit from Vehicle using class Car(Vehicle).
- Add a new attribute seats for number of seats.
- Override show_details() to display additional details.
- Use super().__init__() to call the parent class constructor.


```python
class Vehicle:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def show_details(self):
        print(f"Vehicle: {self.brand} {self.model} \n{self.year}")
```

```python
class Car(Vehicle):
    def __init__(self, brand, model, year, seats):
        # super() infers the current class and self from this method.
        # Python passes self to the parent's __init__ automatically,
        # so we only supply brand, model, and year here.
        super().__init__(brand, model, year)
        self.seats = seats

    def show_details(self):
        super().show_details()
        print(f"{self.seats}")
```

```python
c1 = Car('Jeep', 'Grand Cherokee', 2015, 6)
c1.show_details()
```

```python
v1 = Vehicle('Tesla', 'Roadster', 2028)
v1.show_details()
```

**Multi Level Inheritance**:

**Class ElectricCar**
- Inherits from Car.
- Adds a new attribute: battery_capacity.
- Overrides show_details() to include electric car-specific details.
- Uses super().__init__() to call Car constructor.

```python
class ElectricCar(Car):
    def __init__(self, brand, model, year, seats, battery_capacity):
        super().__init__(brand, model, year, seats)
        self.battery_capacity = battery_capacity

    def show_details(self):
            super().show_details()
            print(f"{self.battery_capacity}")
```

Create Objects of Vehicle, Car, Electric Car class

```python
v1 = Vehicle('Kelly & Rad', 'Raptor', 2015)
v1.show_details()

c1 = Car('Jeep', 'Grand Cherokee', 2022, 6)
c1.show_details()

ec1 = ElectricCar('Tesla', 'Roadster', 2028, 4, '1000mpv')
ec1.show_details()

```

# Polymorphism

**Square Class**
- Create a class named square.
- Define an __init__ method to initialize an instance variable num.
- Define a method area() that calculates and prints the square of num.

**Cube Class**
- Create a class named cube.
- Define an __init__ method to initialize an instance variable num.
- Define a method area() that calculates and prints the cube of num.

Create object for Class Cube & Square and implement the method area()

```python
class Circle:
    def __init__(self, r):
        self.radius = r

    def area(self):
        print(f"Area: {3.14 * r**2}")

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        print(f"Area: {self.length * self.width}")
```

```python
circle1 = Circle(13)
rectangle1 = Rectangle(24,15)

circle1.area()
rectangle1.area()
```

# Abstract class
Define a common Application Program Interface(API) for a set of subclasses



Abstract Class (Vehicle)

This class cannot be instantiated directly.
It enforces that all subclasses must implement the drive() method.
Concrete Subclasses (Car, Truck)

These classes inherit from Vehicle and implement the drive() method.

If a subclass does not implement all @abstractmethods, Python prevents object creation of that subclass.

```python
from abc import ABC, abstractmethod

 # @abstractmethod is a decorator used inside abstract classes to enforce that  child classes must implement the method.
class Vehicle(ABC):
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    @abstractmethod
    def show_details(self):
        print(f"Vehicle: {self.brand} {self.model} \n{self.year}")

class Car(Vehicle):
    def __init__(self, brand, model, year, seats):
        # super() infers the current class and self from this method.
        # Python passes self to the parent's __init__ automatically,
        # so we only supply brand, model, and year here.
        super().__init__(brand, model, year)
        self.seats = seats

    def show_details(self):
        super().show_details()
        print(f"{self.seats}")


class ElectricCar(Car):
    def __init__(self, brand, model, year, seats, battery_capacity):
        super().__init__(brand, model, year, seats)
        self.battery_capacity = battery_capacity

    def show_details(self):
            super().show_details()
            print(f"{self.battery_capacity}")
```
