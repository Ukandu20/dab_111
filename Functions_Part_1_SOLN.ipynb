## Namespace separation

A **namespace** (also called scope) is a region of a program in which **identifiers** have meaning. 

````python
x = 500 

def f(x):
    print(id(x), x)
    x = x + 9
    print(id(x), x)

print(id(x), x)
f(x)
print(id(x), x)
````

## Function calls and definition

The basic syntax for creating a function is

```python
def <function_name>([<parameters>]):
    <statement(s)>
```

To use a function we generally only care about the interface:

- what **arguments** it takes (could be none)
- what **values** it returns and their type

So to call a function we use:

```python
<function_name>([<arguments>])
```

````python
def f(a, b):
    s = a + b
    
f(1, 2)
````

## Side effects

A function creates side effects if changes the calling environment in any way.

````python
my_list = [1, 2, 3]

def f_list(x):
    x[0] = 99
    print(x, id(x))
    
print(my_list, id(my_list))
f_list(my_list)
print(my_list, id(my_list))
````

## `return` statement

Provides a way to modify the calling environment without using side effects. 

A **return** statement serves two purposes:
- it immediately terminates function execution and passes control back to the calling environment
- it provides a way to pass data back to the calling environment (which can be used to modify the environment)

### Exiting a function

````python
# falling off the end

def f():
    print('hi')
    
x = f()
x = print("hi")
id(x)
````

````python
# empty return

def f():
    return
````

````python
# using conditions

def f(a):
    if a>90:
        print("a is greater than 90")
        return 
    else:
        print("a is not greater than 90")
        return 
````

````python
x = f(99)
print(x)
````

````python
# return type None

type(f(9))
````

### Returning data

````python
# immutable but modify calling environment

x = 5

def f(x):
    print(x, id(x))
    x = 10
    print(x, id(x))
    return x

x = f(x)
x
````

````python
# mutable modify calling environment without side effects

def f(x):
    r = []
    for el in x:
        r.append(el**2)
    print(r)
    return r
    
x = [1, 2, 3]

x = f(x)
x
 
````

## <p style="color:DodgerBlue;">EXERCISE</p>

Create a function that will take in the following arguments:
- name of a text file
- whether punctuation should be removed or not
    - make sure this has a default value to not remove puntuation
- whether or not to count words
    - make sure this has a default value to not count words

Your function should:
 - assert that the filename is a string
 - assert that the extension of the filename is ".txt"
 - use `try`/`except` to catch a FileNotFoundError when trying to read in the file
     - if no error, read the text in as a string
 - lower case the text
 - check punctuation argument
     - if True, remove punctuation from provided string
 - check word count argument
     - if True, calculate total word count and unique word count
 - return text as a list split on white space and word counts if necessary
 

````python
import string

def remove_punct(text):
    text_no_punct = ""
    for ch in text:
        if ch not in string.punctuation:
            text_no_punct += ch
            
    return text_no_punct
````

````python
def word_counts(lst):
    total = len(lst)
    unique = len(set(lst))
    return total, unique
````

````python
def get_text_data(fname, punct=False, count_words=False):
    assert isinstance(fname, str), "filename needs to be a string"
    assert fname[-4:]==".txt", "file needs to be a text file"
    try:
        with open(fname) as f:
            data = f.read()
    except FileNotFoundError:
        print("no such file")
        
    data = data.lower()
    
    if punct:
        data = remove_punct(data)

    data = data.split()
        
    if count_words:
        total, unique = word_counts(data)
        return total, unique, data
        
    return data
    
````
