## 1. Introduction to `pathlib`
`pathlib` is part of Python's standard library. The `Path` class represents a filesystem path as an object, making it easier to work with files and folders across operating systems.

```python
from pathlib import Path

p = Path.cwd()
print("Current working directory:", p)
```

### `Path.cwd()` and `Path.iterdir()`
- `Path.cwd()` returns the current working directory.
- `iterdir()` iterates over the immediate files and subdirectories of a folder (not recursively).

```python
for item in p.iterdir():
    print(item)
```

## 2. Understanding File Names and Extensions
A `Path` object has useful properties:

| Property | Meaning | Example for `report.csv` |
|---|---|---|
| `.name` | Filename with extension | `report.csv` |
| `.stem` | Filename without final extension | `report` |
| `.suffix` | Final file extension | `.csv` |

These properties also work on paths that do not currently exist.

```python
sample_path = Path("C:/st_clair/dab_501/okechi_ukandu_lab1.ipynb")
print("Name:", sample_path.name)
print("Stem:", sample_path.stem)
print("Suffix:", sample_path.suffix)
```

```python
print(p.name)
```

```python
for item in p.iterdir():
    print(f"{item.name:<25} | {item.stem:<20} | {item.suffix}")
```

### Raw strings for Windows paths
A raw string prefix (`r`) prevents backslashes from being interpreted as escape sequences such as `\n` and `\t`. This is useful for Windows-style paths.

```python
windows_path = Path(r"C:\Users\Student\Documents\Python")
print(windows_path)
# This is a path example; the directory need not exist on your computer.
```

## 5. Worked Example – Find `.txt` Files with a Loop
Use `iterdir()` and `suffix` to collect filenames from the top-level folder only.

```python
text_files = []
for item in practice.iterdir():
    if item.is_file() and item.suffix == '.txt':
        text_files.append(item.name)
print(sorted(text_files))
```

## 6. Worked Example – `glob()` vs `rglob()`
- `glob("*.txt")` matches `.txt` files in the current folder only.
- `rglob("*.txt")` also searches subfolders.

Results are sorted here for predictable display.

```python
print("Top-level:", sorted(item.name for item in practice.glob("*.txt") if item.is_file()))
print("Recursive:", sorted(str(item.relative_to(practice)) for item in practice.rglob("*.txt") if item.is_file()))
```
