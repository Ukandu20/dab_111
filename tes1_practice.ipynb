Write a Python program that initializes the following nested list: [ [8, 18, 29, 14], [6, 27, 27, 18], [7, 19, 26, 27], [24, 22, 2, 3], [5, 12, 28, 3] ] Find and print the sum of elements at index 2 in each inner list. Expected output: 112



```python
list1 = [[8, 18, 29, 14], [6, 27, 27, 18], [7, 19, 26, 27], [24, 22, 2, 3], [5, 12, 28, 3]]

total = 0

for element in list1:
    total += element[2]

print(total)
```

Write a Python program that creates a tuple with 5 strings: MIpX, APms, iQNG, ddlM, dWOQ. Concatenate them and print the result. Expected output: MIpXAPmsiQNGddlMdWOQ

```python
tuple1 = ('MIpX', 'APms', 'iQNG', 'ddlM', 'dWOQ')
newString = ''

for string in tuple1:
    newString += string

print(newString)
```
