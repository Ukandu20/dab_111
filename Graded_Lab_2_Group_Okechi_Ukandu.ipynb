For example, given the input `[5, 6, 2, 1]` the sums of consecutive elements would be `5 + 6 = 11`, `6 + 2 = 8`, and `2 + 1 = 3`, so the function would return `11`.

- Write a function called `average_str_length` that takes a list of strings as input and returns the average length of the strings in the list. Round the result to the nearest integer.


```python
list2 = ['agddg', 'ae', 'wqwqwefwefdhf']

def conv_to_str(x=[]):
    """ 
        Purpose: 
            converts every element of a list into a string

        Parameters/Arguments: 
            x(list): which is the list of elements of varying datatypes before conversion. It defaults to an empty list

        Return: 
            list: Return a converted list of strings. 
    """
    list_str = []
    assert isinstance(x, list), "x must be a list"
    for element in x:
        element = str(element)
        list_str.append(element)
    return list_str
        
def len_of_str(x=[]):
    """ 
        Purpose:
            Gets the length of each element in a list.
    
        Parameters: 
            x(string-list): which is the list of items of varying datatypes. It defaults to an empty list
            
        Returns:
            list: Returns a list containing the length of every element from the orginal list
            
    """
    str_len = []
    assert isinstance(x, list), "x must be a list"
    for element in conv_to_str(x):
        str_len.append(len(element))
    return str_len

def average_str_length(x=[]):
    """ 
        Purpose:
            Gets the average length of elements in the entire list.
    
        Parameters: 
            x(string-list): which is the list of items of varying datatypes. It defaults to an empty list
            
        Returns:
            int: Returns the approximate average length of all the items in the list.
            
    """
    assert isinstance(x, list), "x must be a list."    
    for value in conv_to_str(x):
        assert isinstance(value, str), "Each item should be a string"
    count = len(x)
    sum = 0

    if count > 0:
        for i in len_of_str(x):
            sum = sum + i
        average = round(sum / count)
    else:
        print("List is empty")
        return 0

    return average

average = average_str_length(list2)
print(f"The average string length is approximately {average}")
    
```

- Create a function named `common_digits` that takes two integers as input and returns the count of common digits between them. For example, for the input 456 and 564, the function should return 3.


```python
def convert_to_str(*args):
    """
        Purpose:
            converts every element into a string, and adds them to a set.
    
        Parameters: 
            *args: which is an integer or a tuple of integers
            
        Returns:
            set: Returns a string-set.
    """
    list_char = {}
    for value in (args):
        assert type(value) is int, "Each value should be an integer"
        list_char = set(str(value))
    return list_char

def common_digits(x,y):
    """
        Purpose:
            Takes two integers and returns the number of common digits between them.
    
        Parameters: 
            x(int): Integer 1
            y(int): Integer 2
            
        Returns:
            int: Returns the value of the total number of common digits
    """
    assert isinstance(x, int), "x must be an integer."
    assert isinstance(y, int), "y must be an integer."
    x = convert_to_str(abs(x))
    y = convert_to_str(abs(y))

    common = x & y
    count = len(common)
    return count
        
common_digits(-32198,-9865)
```

### Question 4

- Create a function called `alternate_merge` that takes two lists and merges them alternately. For example, given the lists [1, 2, 3] and ['a', 'b', 'c'], the function should return [1, 'a', 2, 'b', 3, 'c'].




```python
def alternate_merge(x=[],y=[]):
    """ Purpose:
            Merge two lists by alternating their elements
        
        parameters:
            x(list): list 1
            y(list): list 2
             
        Return:
            list: Return a combined list of alternating elements
    """
    new_list = []
    assert isinstance(x, list), "x must be a list."
    assert isinstance(y, list), "y must be a list."
    min_len = min(len(x), len(y))
    for key,value in enumerate(x):
        for ind,pos in enumerate(y):
            if key == ind:            
                new_list.append(value)
                new_list.append(pos)


    new_list.extend(x[min_len:])
    new_list.extend(y[min_len:])
            
    return new_list

alternate_merge([1], ['a', 'b', 'c', 'd'])
```

### Question 5

- Write a Python function named `nested_sum` that takes a list of lists where each inner list contains integers. The function should return the sum of all numbers in all lists. For example, given `[[1, 2, 3], [3, 4], [5, 6, 7, 8]]`, the function should return `39`.

```python

def nested_sum(x=[]):
    """ 
        Purpose:
            Return the sum of all integers in a list of lists

        Parameters:
            x(list of lists of int): Inner lists containing integers. Defaults to an empty list.

        Returns:
            int: Sum of all the integers across all the inner lists     
    """
    total = 0
    assert isinstance(x, list), "Input must be a list."

    for inner_list in x:
        assert isinstance(inner_list, list), "Each item must be a list."
        for value in inner_list:
            assert type(value) is int, "Each inner list must contain only integers."
        
    for elements in x:
        for element in elements:
            total = total + element
    return total

nested_sum([[1, 2, 3], [3, 4], [5,6, 7, 8]])
```

- Create a function called `unique_characters` that takes in a string and returns `True` if all characters in the string are unique, and `False` otherwise. Ignore spaces. 

```python
def unique_characters(x=''):
    """ Checks if all non-space characters in a string are unique
        Parameters:
            x (str): The string to be examined. Defaults to an empty string
            
        Returns:
            Bool: True if all non-space characters are unique, else false"""
    single_chars = []
    doubles = 0
    assert isinstance(x, str), "Input must be a string."
    x = x.replace(" ", "")
    for i in x:
        if i not in single_chars:
            single_chars.append(i)
        else: 
            doubles = doubles + 1

    if doubles > 0:
        return False
    else:
        return True

unique_characters('Aa.c')
```
