## Question 1 

**Write a Python program that:**

- Initializes a list1 of 10 random integers between 10 and 50 using the random module with a seed 10 (e.g., random.seed(5)).
- Asks the user to enter a number.
- Adds the user-entered number to each element in the list.
- Filters the updated list to keep only those values that are greater than or equal to 30.
- Print the following as output
    - The original list
    - The modified list
    - The final filtered list
      
      
**Note: Both answer below are correct, Submit one answer per question.** 

```python
import random


random.seed(5) # Setting seed ensures reproducibility

list1 = [random.randint(10, 50) for _ in range(10)] # Initialize list of 10 integer numbers with random values between 10 to 50 

add_value = int(input("Enter a number to add: ")) # Get number from user and convert to integer 

modified_list = [num + add_value for num in list1] # Adds the value recieved from user to all numbers in list1

filtered_list = [num for num in modified_list if num >= 30] #Filter list to include only numbers >= 30

# Printing Output
print("Original list:", list1)
print("Modified list:", modified_list)
print("Filtered list:", filtered_list)
```

```python
import random

random.seed(5) # Setting seed ensures reproducibility

list1 = [random.randint(10, 50) for _ in range(10)] # Initialize list of 10 integer numbers with random values between 10 to 50 

add_value = int(input("Enter a number to add: ")) # Get number from user and convert to integer 

modified_list =[] # Initialize a list to store the new values of number after adding add_value

for i in list1: # Iterates through the list1 and adds the value recieved from user to all numbers in list1
    modified_list.append(i+add_value)

filtered_list = [num for num in modified_list if num >= 30] #Filter list to include only numbers >= 30

# Printing Output
print("Original list:", list1)
print("Modified list:", modified_list)
print("Filtered list:", filtered_list)
```
