Copy the `fruit_dict` to `new_fruit_dict`

```python

fruit_dict1 = {'apple': 29, 'banana': 32, 'orange': 244, 'grape': 15}
fruit_dict2 = {'apple': 11, 'kiwi': 32}


fruit_dict1_keys = fruit_dict1.keys()
print(fruit_dict1_keys)


fruit_dict_new = {}

for key,value in fruit_dict2.items():
    if key in fruit_dict1_keys:
        fruit_dict_new[key] = fruit_dict1[key] + fruit_dict2[key]
    else:
        fruit_dict_new.update({key: value})

print(fruit_dict_new)
```
