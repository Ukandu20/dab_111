### Question 14 Classify characters

**Level:** Foundation

For a string, build three lists: alphabetic characters, digit characters and whitespace characters. Preserve duplicates and order within each list. Also print the number of characters that fit none of those groups.

**Requirement:** Use isalpha(), isdigit() and isspace(). Keep digit characters as strings.

**Sample data or result**
```text

Letters: ["A", "b"] | Digits: ["2", "3"]
Whitespace: [" "] | Other count: 1
```

**Additional checks:** Try tabs/newlines and a string containing only punctuation.

````python
# Question 14: Classify characters
# Write and test your solution here.

newWord = input("Enter a Word: ")

letters = []
digits = []
whiteSpaces = []
others = []

for element in newWord:
    if element.isdigit():
        digits.append(element)
    elif element.isalpha():
        letters.append(element)
    elif element == " ":
        whiteSpaces.append(element)
    else:
        others.append(element)

otherCount = len(others)
print(f"Letters: {letters} | Digits: {digits}\nWhitespace: {whiteSpaces} | Other Count: {otherCount}")
````
